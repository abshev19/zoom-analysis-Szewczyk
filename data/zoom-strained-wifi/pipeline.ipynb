```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

fig, ax = plt.subplots(figsize=(10, 8))
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.axis('off')

# Boxes
boxes = [
    ("Raw PCAP Files", 5, 9, 2, 0.6),
    ("Zoom Lua Dissector\n(Wireshark)", 5, 7.5, 2, 0.8),
    ("Packet Parsing\n• Media type\n• Timestamps\n• Sequence #", 5, 5.5, 2, 1.2),
    ("Aggregation", 5, 3, 2, 1.5),
    ("CSV Reports +\nHTML Dashboard", 5, 1, 2, 0.8)
]

for text, x, y, width, height in boxes:
    rect = patches.Rectangle((x-width/2, y-height/2), width, height,
                            linewidth=2, edgecolor='black', facecolor='lightblue',
                            alpha=0.7)
    ax.add_patch(rect)
    ax.text(x, y, text, ha='center', va='center', fontsize=9)

# Arrows
for i in range(len(boxes)-1):
    y1 = boxes[i][2] - boxes[i][4]/2
    y2 = boxes[i+1][2] + boxes[i+1][4]/2
    ax.arrow(5, y1, 0, y2-y1-0.2, head_width=0.1, head_length=0.1, fc='black')

# Aggregation details
agg_details = [
    "• Frame-level (frames.csv)",
    "• Stream-level (streams.csv)",
    "• Temporal (rate.csv, stats.csv)",
    "• Flow-level (flows.csv)"
]
for i, text in enumerate(agg_details):
    ax.text(5, 2.8 - i*0.3, text, ha='center', va='center', fontsize=8)

plt.title("Zoom-Analysis Data Processing Pipeline", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('pipeline_diagram.png', dpi=300, bbox_inches='tight')
plt.show()
```
