```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the frames CSV
df = pd.read_csv("frames.csv")

# Optional: Convert timestamp to float seconds
df["ts_s"] = df["min_ts_s"]


# Group by timestamp and compute average jitter and FPS
metrics_per_time = df.groupby("ts_s")[["jitter_ms", "fps"]].mean().reset_index()

# Save to CSV
metrics_per_time.to_csv("jitter_fps_per_time.csv", index=False)

# Plot jitter
plt.figure()
plt.plot(metrics_per_time["ts_s"], metrics_per_time["jitter_ms"])
plt.xlabel("Timestamp")
plt.ylabel("Jitter (ms)")
plt.title("Jitter Over Time")
plt.show()

# Plot FPS
plt.figure()
plt.plot(metrics_per_time["ts_s"], metrics_per_time["fps"])
plt.xlabel("Timestamp")
plt.ylabel("Frames Per Second")
plt.title("FPS Over Time")
plt.show()
```
